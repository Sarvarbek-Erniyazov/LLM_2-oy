```python
import os
import platform
import psutil
import torch

print("=" * 60)
print("GPU RUNTIME INFORMATION")
print("=" * 60)

print(f"Execution location : {os.getcwd()}")
print(f"Platform           : {platform.platform()}")
print(f"Total RAM          : {psutil.virtual_memory().total / (1024**3):.2f} GB")
print(f"CUDA available     : {torch.cuda.is_available()}")

if torch.cuda.is_available():
    print(f"GPU model          : {torch.cuda.get_device_name(0)}")
    print(f"CUDA version       : {torch.version.cuda}")
    print(f"GPU count          : {torch.cuda.device_count()}")
    
    gpu_memory = torch.cuda.get_device_properties(0).total_memory
    print(f"GPU VRAM           : {gpu_memory / (1024**3):.2f} GB")
    print("Selected device    : GPU")
else:
    print("GPU model          : GPU not detected")
    print("Selected device    : CPU")

print("=" * 60)
```

```python
import os
import platform
import psutil

print("=" * 60)
print("TPU RUNTIME INFORMATION")
print("=" * 60)

print(f"Execution location : {os.getcwd()}")
print(f"Platform           : {platform.platform()}")
print(f"Total RAM          : {psutil.virtual_memory().total / (1024**3):.2f} GB")

try:
    import jax
    
    devices = jax.devices()
    
    print(f"JAX version        : {jax.__version__}")
    print(f"Available devices  : {devices}")
    print(f"Device count       : {len(devices)}")
    print(f"Default backend    : {jax.default_backend()}")
    
    if jax.default_backend() == "tpu":
        print("TPU available      : True")
        print("Selected device    : TPU")
    else:
        print("TPU available      : False")
        print(f"Selected device    : {jax.default_backend().upper()}")

except Exception as error:
    print("TPU available      : Could not verify")
    print(f"Error              : {error}")

print("=" * 60)
```
